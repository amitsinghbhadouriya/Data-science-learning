```python
# converting python objects to json (serialization)
```

```python
# json.dumps()   -> convert Python object to JSON string
import json 
data = {
    "name": "Amit",
    "age": 21,
}
json_string = json.dumps(data)
print(json_string)
```

```python
# json.dump()    -> Write Python object as JSON to a file
import json 
data = {
    "name": "Amit",
    "age": 21,
}
with open("data.json", "w") as file:
    json.dump(data, file, indent=4)
```

```python
# converting JSON to python objects (Deserialization)
```

```python
# json.loads()   -> convert JSON string to python object
import json
json_data = '{"name":"Amit", "age":21}'
python_obj = json.loads(json_data)
print(python_obj)
```

```python
# json.load()    -> Read json data from a file and convert it to python object
import json
with open("data.json", "r") as file:
    python_data = json.load(file)
print(python_data)
```
