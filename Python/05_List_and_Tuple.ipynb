```python
# List
```

```python
num = [5,1,4,3,2]
print(type(num))
print(num)
print(num[0:3])
```

```python
# List Mehtods
```

```python
# Append
num.append(6)
print(num)
```

```python
# Extend
num.extend([7,8])
print(num)
```

```python
# Insert
num.insert(2, "Hey")
print(num)
```

```python
# Remove
num.remove("Hey")
print(num)
```

```python
# Pop
num.pop(3)
print(num)
```

```python
# Count
print(num.count(5))
```

```python
# Sort
num.sort()
print(num)
```

```python
# Reverse
num.reverse()
print(num)
```

```python
# Copy
num.copy()
print(num)
```

```python
# Clear
num.clear()
print(num)
```

```python
# List Comprehension
```

```python
evens = [x for x in range(10) if x % 2 == 0]
print(evens)
```

```python
# Tuple
```

```python
tup = (1,2,3,4,5)
print(type(tup))
print(tup)
print(tup[3])
print(tup[0:3])
```

```python
# Tuple Methods
```

```python
# count
print(tup.count(2))
```

```python
# index
print(tup.index(3))
```

```python
# Tuple comphrension
cubes = (x**3 for x in range(1,11))
print(tuple(cubes))
```
