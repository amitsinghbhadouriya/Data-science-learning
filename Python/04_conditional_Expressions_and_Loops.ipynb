```python
# Conditional Expressions
```

```python
# if-elif-else
x = int(input("Enter a number: "))
if x > 10:
    print("Greater than 10")
elif x == 10:
    print("Equalls to 10")
else:
    print("Smaller than 10")
```

```python
# Match case
day = int(input("Enter a number: "))
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Not Valid!, Enter a number from 1-7")
        
```

```python
# Loops
```

```python
# For loop
list = [2, 4, 6, 8]
for item in list:
    print(item)
```

```python
# While loop
count = 0
while count < 3:
    print(count)
    count += 1
```

```python
# Range Function
for i in range(3):
    print(i)
```

```python
# Break 
count = 0
while count < 10:
    if count == 5:
        break
    print(count)
    count+=1
```

```python
# continue
count = 0
while count < 10:
    if count == 3 or count == 5 or count == 7:
        count+=1
        continue
        
    print(count)
    count+=1
```
