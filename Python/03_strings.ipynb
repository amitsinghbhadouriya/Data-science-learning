```python
# Strings
```

```python
name = "Amit"
print(type(name))
print(name)
```

```python
# String indexing
print(name[2])
```

```python
# String Slicing
name[0:2]
```

```python
# String immutability
name[0] = "S"
name
```

```python
# Escape sequence characters
message = "He is a good\tman.\nHe help others."
print(message)
```

```python
# String Methods
```

```python
# len
print(len(name))
```

```python
# Lower
print(name.lower())
```

```python
# Upper
print(name.upper())
```

```python
# strip
print(name.strip())
```

```python
# replce
print(name.replace("Amit","Abhay"))
```

```python
# split
print(message.split())
```

```python
# join
list = ["hello", "python"]
print(" ".join(list))
```

```python
# Find
print(message.find("good"))
```

```python
# Count
print(message.count("o"))
```

```python
# Startswith
print(name.startswith("A"))
```

```python
# Endswith
print(name.endswith("t"))
```

```python
# isdigit
print(name.isdigit())
```

```python
# isalpha
print(name.isalpha())
```

```python
# isalnum
print(name.isalnum())
```

```python
# String Formatting (F-strings)
name = "Amit"
age = 21
print(f"My name is {name}, and I am {age} years old.")
```
