```python
# Set
```

```python
num = {1,2,3,4,5}
print(type(num))
print(num)
```

```python
# Set Methods
```

```python
# add
num.add(6)
print(num)
```

```python
# update
num.update([7,8,9])
print(num)
```

```python
# copy
nums = num.copy()
print(nums)
```

```python
# remove
num.remove(3)
print(num)
```

```python
# discard
num.discard(3)
print(num)
```

```python
# pop
num.pop()
print(num)
```

```python
# clear
num.clear()
print(num)
```

```python
# Set Operations
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}
```

```python
# Union
set3 = set1.union(set2)
print(set3)
```

```python
# Intersection
set3 = set1.intersection(set2)
print(set3)
```

```python
# Difference
set3 = set1.difference(set2)
print(set3)
```

```python
# Symmetric_difference
set3 = set1.symmetric_difference(set2)
print(set3)
```

```python
# issubset
set3 = set1.issubset(set2)
print(set3)
```

```python
# issuperset
set3 = set1.issuperset(set2)
print(set3)
```

```python
# Dictionary
```

```python
student = {
    "name": "Amit",
    "age": 21,
    "grade": "A",
}
print(type(student))
print(student)
```

```python
# Dictionary Methods
```

```python
# keys
print(student.keys())
```

```python
# values
print(student.values())
```

```python
# items
print(student.items())
```

```python
# copy
print(student.copy())
```

```python
# get
print(student.get("age"))
```

```python
# update
student.update({"age":22})
dict2 = student
print(dict2)
```

```python
# pop
print(student.pop("grade"))
```

```python
# popitem
print(student.popitem())
```

```python
# setdefault
print(student.setdefault("City","Gwalior"))
```

```python
# clear
print(student.clear())
```

```python
# Dictionary comprehension
```

```python
squares = {x: x**2 for x in range(1,11)}
print(squares)
```
