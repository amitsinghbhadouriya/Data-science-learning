```python
# File handling
```

```python
# opening a file
file = open("demo.txt", "r")
lines = file.read()     # Reading a file
print(lines)
file.close()
```

```python
# writing a file
file = open("demo2.txt", "w")
file.write("Hey")
file.close()
```

```python
# Appending a file
file = open("demo.txt", "a")
file.write("\nwhat are you learning in python")
file.close()
```

```python
# With statement
with open("demo.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
# Checking if a file exists or not
import os
if os.path.exists("demo.txt"):
    print("File exists!")
else:
    print("File not found")
```

```python
# Deleting a file
import os 
if os.path.exists("test.txt"):
    os.remove("test.txt")
    print("File deleted")
else:
    print("File does not exist")
```

```python
# Working with binary files
with open("binary.jpg", "rb") as file:
    data = file.read()
    print(data)
    
```
