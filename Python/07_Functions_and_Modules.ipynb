```python
# Functions
```

```python
def greet():
    return f"Hello, world!"

print(greet())
```

```python
# Function with parameters
def greet(name):
    return f"Hello, {name}"

print(greet("Amit"))
```

```python
def sum(a,b):
    return a+b

print(sum(4,5))
```

```python
# lambda function
sq = lambda x: x**2
print(sq(5))
```

```python
# Recursion
def feb(n):
    if n == 0 and n == 1:
        return n
    return (n-2)+(n-1)

print(feb(5))
```

```python
# Local variable and Global variable
x = 5    # global variable
def my_fun():
    x = 2   # local variable
    print(x)

my_fun()
print(x)
```

```python
# Docstrings
def add(a,b):
    ''' Returns sum of two number. '''
    return a+b

print(add.__doc__)
```

```python
# Modules
import math
print(math.sqrt(16))     # Square root
print(math.pow(4,2))     # Square
```
