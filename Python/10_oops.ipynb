```python
class car:
    def __init__(self, brand, model):     # Constructor
        self.brand = brand
        self.model = model

    def display_info(self):
        return f"{self.brand} {self.model}"

car1 = car("Toyota", "camry")
print(car1.display_info())
```

```python
# Encapsulation
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks   # private variable

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if marks >= 0 and marks <= 100:
            self.__marks = marks
        else:
            return "Invalid marks"


student = Student("Amit", 85)

print(student.name)
print(student.get_marks())

student.set_marks(90)
print(student.get_marks())
```

```python
# Inheritance
```

```python
# Single Inheritance
class Animal:
    def speak(self):
        return "Animal makes a sound"

class Dog(Animal):
    def sound(self):
        return "Bark"

dog = Dog()
print(dog.sound())
print(dog.speak())
```

```python
# Multilevel Inheritace
class Animal:
    def eat(self):
        return "Animal is eating"

class Dog(Animal):
    def bark(self):
        return "Dog is barking"

class Puppy(Dog):
    def play(self):
        return "Puppy is playing"

p = Puppy()
p.eat()    
p.bark()   
p.play()   
```

```python
# Multiple Inheritance
class Father:
    def skills(self):
        return "Father: Driving"

class Mother:
    def hobbies(self):
        return "Mother: Cooking"

class Son(Father, Mother):
    def study(self):
        return "Son: Studying"

s = Son()

s.skills()
s.hobbies()
s.study()
```

```python
# Hierarichial Inheritance
class Animal:
    def eat(self):
        return "Animal is eating"

class Dog(Animal):
    def bark(self):
        return "Dog is barking"

class Cat(Animal):
    def meow(self):
        return "Cat is meowing"

dog = Dog()
dog.eat()
dog.bark()

cat = Cat()
cat.eat()
cat.meow()
```

```python
# Hybrid Inheritance
class Animal:
    def eat(self):
        return "Animal is eating"

class Dog(Animal):
    def bark(self):
        return "Dog is barking"

class Cat(Animal):
    def meow(self):
        return "Cat is meowing"

class Pet(Dog, Cat):
    def play(self):
        return "Pet is playing"

p = Pet()
p.eat()
p.bark()
p.meow()
p.play()
```

```python
# Polymorphism
```

```python
# Method overriding
class Bird:
    def fly(self):
        return "Birds can fly"

class Penguin(Bird):
    def fly(self):
        return "Penguins cannot fly"

bird = Bird()
penguin = Penguin()
print(bird.fly())
print(penguin.fly())
```

```python
# Duck Typing
class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("Cat meows")

def make_sound(animal):
    animal.sound()

dog = Dog()
cat = Cat()
make_sound(dog)
make_sound(cat)
```

```python
# Operator overloading
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __add__(self, other):
        return self.marks + other.marks

s1 = Student(80)
s2 = Student(90)

print(s1 + s2)
```

```python
# Abstraction
```

```python
from abc import ABC, abstractmethod
class shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Square(shape):
    def __init__(self, side):
        self.side = side
        
    def area(self):
        return self.side * self.side

square = Square(4)
print(square.area())
```

```python
# Magic Methods (Dunder Methods)
```

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"Book: {self.title}"

    def __len__(self):
        return self.pages

book = Book("Python Basics", 300)
print(str(book))
print(len(book))
```

```python
# class vs Static Methods
```

```python
class Example:
    class_var = "I am a class variable"

    def instance_method(self):
        return "Instance Method"

    @classmethod
    def class_method(cls):
        return cls.class_var

    @staticmethod
    def static_method():
        return "Static Mehtod"

obj = Example()
print(obj.instance_method())
print(Example.class_method())
print(Example.static_method())
```

```python
# Decorators
```

```python
def my_decorator(func):
    def wrapper():
        print("Hello")
        func()
        print("Hey")
    return wrapper

@my_decorator
def say_hello():
    print("Hello world!")

say_hello()
```

```python
# using Decorators with Arguments
def repeat(n):
    def decorator(func):
        def wrapper(a):
            for _ in range(n):
                func(a)
        return wrapper
    return decorator

@repeat(3)
def greet(name):
    print(f"Hello, {name}")

greet("world")
```

```python
# Getter and Setter
```

```python
class person:
    def __init__(self, name):
        self._name = name

    def get_name(self):
        return self._name

    def set_name(self, new_name):
        self._name = new_name

p = person("Amit")
print(p.get_name())
p.set_name("Singh")
print(p.get_name())
```

```python
# using @property
```

```python
class Person:
    def __init__(self, name):
        self._name = name

    @property 
    def name(self):     # Getter
        return self._name

    @name.setter
    def name(self, new_name):    # Setter
        self._name = new_name

p = Person("Amit")
print(p.name)
p.name = "singh"
print(p.name)
```
