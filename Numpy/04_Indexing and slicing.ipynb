```python
import numpy as np

arr = np.array([10,20,30,40,50])
```

```python
# Indexing
print(arr[0])
print(arr[-1])
```

```python
# Slicing
print(arr[0:2])
print(arr[:3])
print(arr[::2])
```

```python
sliced = arr[1:4]
sliced[0] = 999
print(arr)
```
