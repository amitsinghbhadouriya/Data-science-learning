```python
import numpy as np

arr = np.array([2,3,5,6,4,9,7])
arr1 = [1, 2, 3, 4, 5]
arr2 = [2, 4, 6, 8, 10]
print(f"Original array: {arr}")
```

```python
# Mean (average)
print(f"The mean of this array is: {np.mean(arr)}")
```

```python
# Standard Deviation
print(f"The standard deviation of this array is: {np.std(arr)}")
```

```python
# Variance
print(f"The variance of this array is: {np.var(arr)}")
```

```python
# Minimum
print(f"The minimum value of this array is: {np.min(arr)}")
```

```python
# Maximum
print(f"The maximum value of this array is: {np.max(arr)}")
```

```python
# Sum
print(f"The sum of this array is: {np.sum(arr)}")
```

```python
# Product
print(f"The product of this array is: {np.prod(arr)}")
```

```python
# Median
print(f"The median of this array is: {np.median(arr)}")
```

```python
# Percentile
print(f"The percentile of this array is: {np.percentile(arr,4)}")
```

```python
# Index of the minimum
print(f"The index of the minimum value of this array is: {np.argmin(arr)}")
```

```python
# Index of the maximum
print(f"The index of the maximum value of this array is: {np.argmax(arr)}")
```

```python
# Correlation coefficient
print(f"The correlation coefficient of two arrays are: {np.corrcoef(arr1,arr2)}")
```

```python
# Unique elements
print(f"The unique elements of this array are: {np.unique(arr)}")
```

```python
# n-th differences
print(f"The n-th difference of this array is: {np.diff(arr)}")
```

```python
# Cumulative sum
print(f"The cumulative sum of this array is: {np.cumsum(arr)}")
```

```python
# Evenly spaced numbers
print(f"The evenly spaced numbers over a specified interval are: {np.linspace(0,10,5)}")
```

```python
# natural logarithm
print(f"The natural logarithm of this array is: \n{np.log(arr)}")
```
