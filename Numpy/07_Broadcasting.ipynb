```python
# Looping over arrays in python

import numpy as np

arr = np.array([1,2,3,4,5])
result = []
# Using a loop to square each element
for num in arr:
    result.append(num**2)

print(result)
```

```python
# Vectorized operation
result = arr**2
print(result)
```

```python
# Broadcasting with scalar
result = arr + 10
print(result)
```

```python
# Broadcasting with two arrays
arr1 = np.array([1,2,3])
arr2 = np.array([10,20,30])
result = arr1 + arr2
print(result)
```

```python
# Broadcasting a 2D array and a 1D array
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([1,2,3])
result = arr1 + arr2
print(result)
```
