```python
# Creating DataFrames by different types
import pandas as pd
```

```python
# From Python lists
data = [
    ["Amit", 21],
    ["Karan", 25],
    ["Naitik", 28]
]
df = pd.DataFrame(data, columns=["Name", "Age"])
print(f"Creating dataframe through python lists: \n{df}")
```

```python
# From Dictionary of lists
data = {
    "Name": ["Amit", "Dev", "Aman"],
    "Age": [21, 23, 18]
}
df = pd.DataFrame(data)
print(f"Creating dataframe through dictionary of lists: \n{df}")
```

```python
# From NumPy Arrays
import numpy as np

arr = np.array([[1,2], [3,4]])
df = pd.DataFrame(arr, columns=["A", "B"])
print(f"Creating dataframe through NumPy arrays: \n{df}")
```

```python
# From CSV files

df = pd.read_csv("data.csv")
print(f"Creating dataframe through CSV file: \n{df}")

df2 = pd.read_csv("data.csv", usecols=["name"])
print(f"\nCreating dataframe through CSV file with using a usecols method: \n{df2}")
```

```python
# From Excel file
df = pd.read_excel("data.xlsx")
print(f"Creating dataframe through excel file: \n{df}")
```

```python
# From JSON file
df = pd.read_json("data.json")
print(f"Creating dataframe through json file: \n{df}")
```

```python
# From sql databases
import sqlite3

conn = sqlite3.connect("mydb.sqlite")
df = pd.read_sql("SELECT * FROM data", conn)

print(f"Creating dataframe through sql file: \n{df}")

conn.close()
```
