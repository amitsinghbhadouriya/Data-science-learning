```python
# Series - 1D labeled Array  (like a list)
import pandas as pd

s = pd.Series([10,20,30,40])
print(f"Result of series: \n{s}")
```

```python
# Custom index
s = pd.Series([10,20,30], index = ["a", "b", "c"])
print(f"Customized indexing in series: \n{s}")
```

```python
# DataFrame - 2D labeled table (like a dictionary of series)
data = {
    "name": ["Amit", "Naitik", "Karan"],
    "age": [21, 28, 25],
    "city": ["Gwalior", "Delhi", "Vanarasi"]
}
df = pd.DataFrame(data)
print(f"Result of dataframe: \n{df}")
```

```python
# Index
print(f"Row labels in this dataframe is: \n{df.index}")
# Labels
print(f"\nColumn labels in this dataframe is: \n{df.columns}")
```

```python
# Changing index and labels value
df.index = ["a", "b", "c"]
df.columns = ["Name", "Age", "City"]
print(f"Updated dataframe: \n{df}")
```
