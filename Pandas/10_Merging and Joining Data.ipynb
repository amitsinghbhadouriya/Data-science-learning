```python
import pandas as pd
```

```python
Data1 = {
    "EmpID": [1, 2, 3],
    "Name": ["Amit", "Naitik", "Shivam"],
    "DeptID": [10, 20, 30]
}
employees = pd.DataFrame(Data1)
employees
```

```python
Data2 = {
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
}
departments = pd.DataFrame(Data2)
departments
```

```python
# Merges
```

```python
# Inner Join
pd.merge(employees, departments, on="DeptID")
```

```python
# Left Join
pd.merge(employees, departments, on="DeptID", how="left")
```

```python
# Right Join
pd.merge(employees, departments, on="DeptID", how="right")
```

```python
# Outer Join
pd.merge(employees, departments, on="DeptID", how="outer")
```

```python
# concatenating DataFrames
```

```python
# Vertical (Rows)
df1 = pd.DataFrame({"Name": ["Amit", "Naitik"]})
df2 = pd.DataFrame({"Name": ["Rohan", "Advait"]})

pd.concat([df1, df2])
```

```python
# Horizontal (Columns)
df1 = pd.DataFrame({"ID": [1, 2]})
df2 = pd.DataFrame({"Score": [90, 80]})

pd.concat([df1, df2], axis=1)
```
