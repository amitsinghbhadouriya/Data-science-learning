```python
import pandas as pd

url =  "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
print(df)
```

```python
# Selecting rows and columns
```

```python
# Selecting columns
print(f"Selecting entire single column: \n{df['total_bill']}")
print(f"\nSelecting multiple columns: \n{df[['sex','smoker']]}")
```

```python
# Selecting rows by index
print(f"Selecting first row by label: \n{df.loc[0]}")
print(f"\nSelecting first row by position: \n{df.iloc[0]}")
```

```python
# Selecting specific rows and columns
print(f"Select value at row x and column y: \n{df.loc[0, 'day']}")
print(f"\nSelect value at row x and column at index y: \n{df.iloc[0, 5]}")

# through slicing
print(f"\nselect value through slicing : \n{df.loc[0:2, ['sex','time']]}")
print(f"\nselect value through slicing by position : \n{df.iloc[0:2, 0:2]}")
```

```python
# Using .at and .iat
print(f"Access single element by fast label based: \n{df.at[0,'size']}")
print(f"\nAccess single element by fast position based: \n{df.iat[0,4]}")
```
