```python
df.drop_duplicates()
```

```python
# string operation with .str
df["customer_name"].str.lower()
```

```python
df["city"].str.contains("delhi",case=False)
```

```python
# Type Conversions with .astype()
```

```python
df["rating"] = df["rating"].fillna(0).astype(int)
```

```python
df["order_date"] = pd.to_datetime(df["order_date"], format="%d-%m-%Y")
```

```python
df.dtypes
```

```python
# Applying functions
```

```python
# .apply() -> Apply any function to rows or columns
df["amount rating"] = df["amount"].apply(lambda x: 5 if x >= 25000 else 4)
```

```python
df[["amount", "amount rating"]].head(10)
```

```python
# .map() -> element wise mapping for series 
df["payment_code"] = df["payment_method"].map({
    "UPI": 1,
    "Credit Card": 2,
    "Debit Card": 3,
    "Cash": 4,
    "Net Banking": 5
})
```

```python
df[["payment_method", "payment_code"]].head(10)
```

```python
# .replace() -> Replace sepecific values
df["payment_method"] = df["payment_method"].replace({
    "upi": "UPI",
    "Upi": "UPI",
    "credit card": "Credit Card",
    "debit card": "Debit Card"
})
```
