```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
        "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})
```

```python
# .groupby() function
df.groupby("Department")["Salary"].mean()
```

```python
# Common Aggrregation function
```

```python
# Mean
print(f"Mean Function: \n{df.groupby('Team')['Salary'].mean()}\n\n")

# Sum 
print(f"Sum Function: \n{df.groupby('Team')['Salary'].sum()}\n\n")

# Count
print(f"Count Function: \n{df.groupby('Team')['Salary'].count()}\n\n")

# Min
print(f"Min Function: \n{df.groupby('Team')['Salary'].min()}\n\n")

# Max
print(f"Max Function: \n{df.groupby('Team')['Salary'].max()}\n\n")
```

```python
# Custom aggregations with .agg()
df.groupby("Team")["Salary"].agg(["mean", "max", "min"])
```

```python
df.groupby("Team")["Salary"].agg(
    avg_score="mean",
    high_score="max"
)
```

```python
df.groupby("Team").agg({
    "Salary": "mean",
    "Age": "max"
})
```

```python
# .transform() example
df["Team Avg"] = df.groupby("Team")["Salary"].transform("mean")
```

```python
# .filter() example
df.groupby("Team").filter(lambda x: x["Salary"].mean() > 80)
```
