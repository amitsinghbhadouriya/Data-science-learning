```python
import pandas as pd

data = {
    "Name" : ["Amit", "Dev", "Nakul", "Rohan"],
    "Maths" : [85, 90, 78, 74],
    "Science" : [88, 92, 80, 89],
    "English" : [92, 89, 94, 86],
}
```

```python
df = pd.DataFrame(data)
df
```

```python
# Using Melt
melted = pd.melt(
    df,
    id_vars=['Name'],
    var_name='Subject',
    value_name='Marks'
)
melted
```

```python
# using Pivot
pivoted = melted.pivot(
    index='Name',
    columns='Subject',
    values='Marks',
    aggfunc = 'mean'
).resetindex()

pivoted
```
