```python
import pandas as pd

data = {
    'Name' : ['Dev', 'Amit', 'Shubham', 'Jeet'],
    'age' : [21, 20, 22, 19],
    'marks' : [98, 78, 89, 95],
}
```

```python
df = pd.DataFrame(data)
df
```

```python
# Sort by values
```

```python
df.sort_values("Name")
```

```python
# Ranking data
```

```python
df['Rank'] = df['marks'].rank(ascending=False)
print("Ranking Data:")
df
```

```python
# Renaming Columns
```

```python
df = df.rename(columns={'Name': 'Student_Name', 'marks': 'Score'})
print("Renamed Columns:")
df
```

```python
# Renaming Index
```

```python
df = df.rename(index={0: 'A', 1: 'B', 2: 'C', 3: 'D'})
print("Renamed Index:")
df
```

```python
# Changing column order
```

```python
df = df[['Student_Name', 'Score', 'age', 'Rank']]
print("Changed Column Order:")
df
```
