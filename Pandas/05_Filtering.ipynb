```python
import pandas as pd

url =  "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
print(df)
```

```python
# Simple condition
print(df[df["size"]>4])
```

```python
# Multiple conditions
print(f"Using And operator: \n{df[(df["size"]>4) & (df["sex"] == "Male")]}")
print(f"\nUsing OR operator: \n{df[(df["smoker"] == "Yes") | (df["sex"] == "Female")]}")
```

```python
# Query with .query()
print(df.query("tip > 3 and smoker == 'Yes' and sex == 'Female'"))
```

```python
col = "total_bill"
print(df.query(f"{col}>2"))
```
