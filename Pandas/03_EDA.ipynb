```python
# Exploratory Data Analysis
```

```python
import pandas as pd
url =  "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
print(df)
```

```python
# Head
print(f"To Check the first 5 row of dataframe: \n{df.head()}")
```

```python
# Tail
print(f"To check the last 5 row of dataframe: \n{df.tail()}")
```

```python
# info
print(f"To find the information about the dataframe:")
df.info()
```

```python
# Describe
print(f"To find the stats and description of the dataframe:")
df.describe()
```

```python
# Columns
print(f"To check the list of column names in a dataframe:")
df.columns
```

```python
# shape
print(f"To find the shape of a dataframe: ")
df.shape
```
