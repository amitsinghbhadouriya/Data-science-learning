```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
print(sns.get_dataset_names())
```

```python
tips = sns.load_dataset('tips')
print(tips.head())
```

```python
# Line plot
sns.lineplot(
    x="total_bill",
    y="tip",
    data=tips
)
plt.title("Line Plot Example")
plt.show()
```

```python
# Scatter Plot
sns.scatterplot(
    x="total_bill",
    y="tip",
    data=tips,
    hue="time"
)
plt.title("Scatter Plot with Color by Time")
plt.show()
```

```python
# Bar Plot
sns.barplot(
    x="day",
    y="total_bill",
    data=tips,
)
plt.title("Average bill per day")
plt.show()
```

```python
# Box plot
sns.boxplot(
    x="day",
    y="total_bill",
    data=tips
)
plt.title("Boxplot of Total Bill per Day")
plt.show()
```

```python
# Heatmap
flights = sns.load_dataset("flights")

pivot_table = flights.pivot(
    index="month",
    columns="year",
    values="passengers"
)

sns.heatmap(pivot_table, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Heatmap of Passengers")
plt.xlabel("Year")
plt.ylabel("Month")
plt.show()
```
