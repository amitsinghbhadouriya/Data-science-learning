```python
import matplotlib.pyplot as plt
```

```python
years = [2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]
runs = [3568, 3126, 3265, 2852, 1308, 1369, 1567, 2632, 1343, 1331]
```

```python
plt.bar(years,runs)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Virat Kholi's Yearly Runs")
plt.show()
```

```python
# Comarasion
import numpy as np
kohli = [3568, 3126, 3265, 2852, 1308, 1369, 1567, 2632, 1343, 1331]
klRahul = [1311, 964, 1406, 1914, 1517, 1484, 1376, 1477, 1044, 1719]
Rohit = [1838, 2126, 2090, 2846, 643, 1727, 1263, 2001, 1442, 1068]

x = np.arange(len(years))
width = 0.25
```

```python
# ploting side by side bars
plt.bar(x - width, kohli, width=width, label="Virat Kohli")
plt.bar(x, klRahul, width=width, label="KL Rahul")
plt.bar(x + width, Rohit, width=width, label="Rohit Sharma")

plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Run Comparison")
plt.xticks(x, years)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Horizontal Bar Charts
players = ["Kholi", "klRahul", "Rohit", "Rishab"]
runs_syrs = [500+700+1100+1500+1800, 500+200+900+1400+1600, 400+600+500+800+1100, 
300+600+800+1100+900]

# plotting barh()
plt.barh(players, runs_syrs, color="skyblue")
plt.xlabel("Total Runs in First 5 years")
plt.title("First 5-Year Performance of Indian Batsman")
plt.tight_layout()
plt.show()
```

```python
# Adding value labels with plt.txt()
players = ["Virat", "klRahul", "Rohit"]
runs = [1800, 1300, 1600]

plt.figure(figsize=(8,5))
plt.bar(players, runs, color="Red")

for i in range(len(players)):
    plt.text(i, runs[i] + 50, str(runs[i]), ha='center')
    
plt.ylabel("Runs")
plt.title("Runs Scored by Players")
plt.tight_layout()
plt.show()
```
