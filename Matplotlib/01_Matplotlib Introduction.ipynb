```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1,2,3], [4,5,6])
plt.show()
```

```python
# Real Example
years = [2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]
runs = [3568, 3126, 3265, 2852, 1308, 1369, 1567, 2632, 1343, 1331]
plt.plot(years, runs)
plt.show()
```

```python
# Adding labels
plt.plot(years,runs)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Virat Kholi's Yearly Runs")
plt.show()
```

```python
# Multiple lines
klRahul = [1311, 964, 1406, 1914, 1517, 1484, 1376, 1477, 1044, 1719]
Rohit = [1838, 2126, 2090, 2846, 643, 1727, 1263, 2001, 1442, 1068]

plt.plot(years, klRahul, label="Kl Rahul")
plt.plot(years, Rohit, label="Rohit Sharma")

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Performance Comparision")
plt.legend()
plt.show()
```

```python
# Using Format Strings
plt.plot(years, klRahul, 'ro--', label="Kl Rahul")
plt.plot(years, Rohit, 'g^:', label="Rohit Sharma")
plt.legend()
plt.show()
```

```python
# Color and line style
plt.plot(years, klRahul, color='blue', linestyle='-.', label="KL Rahul")
plt.plot(years, Rohit, color='orange', linestyle='--', label="Rohit Sharma")
plt.plot(years, runs, color='pink', label="Virat Kholi")
plt.legend()
plt.show()
```

```python
# Line Width and Layout Tweaks
plt.plot(years, klRahul, linewidth=2, label="KL Rahul")
plt.plot(years, Rohit, linewidth=3, label="Rohit Sharma")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.plot(years, klRahul, linewidth=2, label="KL Rahul")
plt.plot(years, Rohit, linewidth=3, label="Rohit Sharma")
plt.style.use("seaborn-v0_8-bright")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# XKCD Comic Style
with plt.xkcd():
    plt.plot(years, klRahul, label="KL Rahul")
    plt.plot(years, Rohit, label="Rohit Sharma")

    plt.title("Epic Battle of the Batsman")
    plt.xlabel("Years")
    plt.ylabel("Runs")

    plt.legend()
    plt.show()
```

```python
# Visualizing tons of data

import numpy as np

for i in range(50):
    plt.plot(np.random.rand(100), linewidth=1)

plt.title("Too much data can be confusing")
plt.grid()
plt.tight_layout()
plt.show()
```
