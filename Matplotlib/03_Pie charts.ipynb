```python
import matplotlib.pyplot as plt
```

```python
plt.style.use("ggplot")
```

```python
labels = ["Virat", "Rohit", "klRahul", "Rishab"]
runs = [28527, 20427, 10094, 5805]

plt.title("Career Runs of Indian Batsman")
plt.pie(runs, labels=labels)
plt.show()
```

```python
# custom Colors
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99'] 
plt.pie(runs, labels=labels, colors=colors)
plt.show()
```

```python
# Highlighting slice
explode = [0, 0, 0, 0.2]

plt.pie(runs, labels=labels, explode=explode, colors=colors)
plt.title("Exploded Pie Example")
plt.show()
```

```python
# Add Shadow
plt.pie(
    runs,
    labels=labels,
    explode=explode,
    colors=colors,
    shadow=True
)
plt.show()
```

```python
# Start Angle
plt.pie(
    runs,
    labels=labels,
    explode=explode,
    colors=colors,
    shadow=True,
    startangle=180
)
plt.show()
```

```python
# Show Percentages
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    explode=explode,
    autopct='%1.1f%%',
    startangle=180
)
plt.title("Carrer Run Share")
plt.show()
```

```python
# Crowded Pie Chart
languages = ["Python", "Java", "C++", "JavaScript", "C#", "Ruby", "Go", "Rust", "Swift", "PHP"]
usage = [30, 20, 10, 10, 7, 5, 4, 3, 2, 1]

plt.pie(
    usage,
    labels=languages,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Programming Language Usage (Crowded Example)")
plt.show()
```
